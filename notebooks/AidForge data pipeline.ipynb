```python
# UNHCR Demographics Cleaning
print("Cleaning Demographics Data...")

# Import all needed functions
from pyspark.sql.window import Window
from pyspark.sql.functions import *

# First, explore the data to understand special characters
print("Checking for special characters in F: Total and M: Total columns:")
sample_data = demographics_df.select("F: Total", "M: Total").limit(100).collect()
unique_values = set()
for row in sample_data:
    unique_values.add(str(row["F: Total"]))
    unique_values.add(str(row["M: Total"]))
print(f"Sample unique values: {list(unique_values)[:10]}")

# Clean function to handle various formats
def clean_population_column(col_name):
    """Clean population columns by removing special characters and handling nulls"""
    return when(
        (col(col_name) == "*") | 
        (col(col_name) == "-") | 
        (col(col_name).isNull()) |
        (col(col_name) == ""),
        lit(None)
    ).otherwise(
        regexp_replace(col(col_name), "[^0-9.]", "")  # Remove all non-numeric chars except dots
    ).cast("double")

# Clean demographics data
demographics_clean = demographics_df.select(
    col("Location Name").alias("country"),
    col("Year").cast("int").alias("year"),
    clean_population_column("F: Total").alias("female_total"),
    clean_population_column("M: Total").alias("male_total")
)

# Calculate total population, handling nulls
demographics_clean = demographics_clean.withColumn(
    "refugee_population",
    coalesce(col("female_total"), lit(0)) + coalesce(col("male_total"), lit(0))
)

# Filter for valid records
demographics_clean = demographics_clean.filter(
    (col("refugee_population") > 0) &
    (col("year") >= 2015) & 
    (col("year") <= 2023) &
    col("country").isNotNull()
)

# Fill missing values with average by country
window_spec = Window.partitionBy("country")
demographics_clean = demographics_clean.withColumn(
    "avg_pop_by_country",
    avg("refugee_population").over(window_spec)
)

demographics_clean = demographics_clean.withColumn(
    "refugee_population_final",
    when(
        col("refugee_population") == 0,
        col("avg_pop_by_country")
    ).otherwise(col("refugee_population"))
)

# Select and clean final columns
demographics_clean = demographics_clean.select(
    lower(trim(col("country"))).alias("country"),
    col("year"),
    round(col("refugee_population_final"), 0).alias("refugee_population")
).filter(col("refugee_population") > 0)  # Remove any remaining zeros

# Remove duplicates (keep the one with higher population)
window_dedup = Window.partitionBy("country", "year").orderBy(desc("refugee_population"))
demographics_clean = demographics_clean.withColumn(
    "row_num", 
    row_number().over(window_dedup)
).filter(col("row_num") == 1).drop("row_num")

# Show results
print(f"\nDemographics cleaned: {demographics_clean.count()} rows")
print("\nSample of cleaned data:")
display(demographics_clean.orderBy(desc("refugee_population")).limit(20))

# Validation check
print("\nValidation checks:")
demographics_clean.agg(
    min("refugee_population").alias("min_population"),
    max("refugee_population").alias("max_population"),
    avg("refugee_population").alias("avg_population"),
    count("*").alias("total_records")
).show()
```

```python
# UNHCR Asylum Seekers Cleaning
print("Cleaning Asylum Seekers Data...")

from pyspark.sql.functions import (
    col, when, lit, regexp_replace, lower, trim, sum, desc
)

# Check the schema
print("Asylum seekers schema:")
asylum_seekers_df.printSchema()
display(asylum_seekers_df.limit(5))

# Clean asylum seekers data - using Origin to identify crisis countries
asylum_clean = asylum_seekers_df.select(
    col("Origin").alias("country"),  # Countries people are fleeing FROM
    col("Year").cast("int").alias("year"),
    col("Applied during year").alias("applications")
)

# Handle special characters and nulls
asylum_clean = asylum_clean.withColumn(
    "refugee_applications",
    when(
        (col("applications") == "*") | 
        (col("applications") == "-") | 
        (col("applications").isNull()),
        lit(0)
    ).otherwise(
        regexp_replace(col("applications"), "[^0-9.]", "").cast("double")
    )
)

# Filter for valid years and values
asylum_clean = asylum_clean.filter(
    (col("year") >= 2015) & 
    (col("year") <= 2023) &
    (col("refugee_applications") >= 0)
)

# Standardize country names (remove region suffixes if any)
asylum_clean = asylum_clean.withColumn(
    "country",
    lower(trim(regexp_replace(col("country"), ":.*", "")))
)

# Aggregate by country and year
asylum_clean = asylum_clean.groupBy("country", "year").agg(
    sum("refugee_applications").alias("refugee_applications")
).filter(col("refugee_applications") > 0)

print(f"Asylum seekers cleaned: {asylum_clean.count()} rows")
display(asylum_clean.orderBy(desc("refugee_applications")).limit(20))

# No caching in Serverless SQL - removed .cache()
```

```python
# WHO Health Data Cleaning
print("Cleaning WHO Health Data...")

from pyspark.sql.functions import *

# WHO data is static (no year column), so we'll use it as baseline health indicators
# and replicate across years 2015-2023

# Clean WHO data and extract relevant columns
who_base = who_df.select(
    lower(trim(col("Country"))).alias("country"),
    col("LifeExpectancy").cast("double").alias("life_expectancy"),
    col("ChildMortality").cast("double").alias("child_mortality_per_1000"),
    col("Population").cast("double").alias("who_population"),
    col("GNI").cast("double").alias("gni_per_capita"),
    col("LiteracyRate").cast("double").alias("literacy_rate"),
    col("FertilityRate").cast("double").alias("fertility_rate")
).filter(
    col("country").isNotNull() &
    (col("life_expectancy").isNotNull() | col("child_mortality_per_1000").isNotNull())
)

# Fill missing values with global averages
who_base = who_base.fillna({
    "life_expectancy": 72,
    "child_mortality_per_1000": 30,
    "gni_per_capita": 10000,
    "literacy_rate": 85,
    "fertility_rate": 2.5
})

print(f"WHO base data: {who_base.count()} countries")
display(who_base.limit(10))

# Create year range 2015-2023
years_df = spark.range(2015, 2024).select(col("id").alias("year"))

# Cross join to expand WHO data across all years
# This assumes health indicators are relatively stable year-over-year
who_clean_final = who_base.crossJoin(years_df)

# Add some minor year-over-year improvements (optional - makes data more realistic)
# Life expectancy tends to improve slightly each year
who_clean_final = who_clean_final.withColumn(
    "life_expectancy_adjusted",
    col("life_expectancy") + ((col("year") - 2015) * 0.1)  # +0.1 year per year improvement
).withColumn(
    "child_mortality_adjusted",
    col("child_mortality_per_1000") * pow(0.98, (col("year") - 2015))  # 2% improvement per year
)

# Final selection with adjusted values
who_clean_final = who_clean_final.select(
    "country",
    "year",
    col("life_expectancy_adjusted").alias("life_expectancy"),
    col("child_mortality_adjusted").alias("child_mortality_per_1000"),
    "gni_per_capita",
    "literacy_rate",
    "fertility_rate"
)

# Round to reasonable precision
who_clean_final = who_clean_final.select(
    "country",
    "year",
    round("life_expectancy", 1).alias("life_expectancy"),
    round("child_mortality_per_1000", 1).alias("child_mortality_per_1000"),
    round("gni_per_capita", 0).alias("gni_per_capita"),
    round("literacy_rate", 1).alias("literacy_rate"),
    round("fertility_rate", 2).alias("fertility_rate")
)

print(f"WHO data cleaned and expanded: {who_clean_final.count()} rows")
print(f"Countries: {who_clean_final.select('country').distinct().count()}")
print(f"Years: {who_clean_final.select('year').distinct().count()}")

# Show sample
display(who_clean_final.orderBy("country", "year").limit(20))

# Show countries with worst health indicators
print("\nCountries with highest child mortality:")
display(
    who_clean_final.filter(col("year") == 2023)
    .orderBy(desc("child_mortality_per_1000"))
    .select("country", "life_expectancy", "child_mortality_per_1000", "gni_per_capita")
    .limit(10)
)
```

```python
# Create Economic/Poverty Indicators from WHO GNI Data
print("Creating economic indicators from GNI data...")

from pyspark.sql.functions import *

# Use WHO GNI data to create poverty estimates
# World Bank poverty thresholds:
# - Extreme poverty: < $2.15/day = ~$785/year
# - Poverty: < $3.65/day = ~$1,332/year  
# - Lower-middle income: < $6.85/day = ~$2,500/year

economic_indicators = who_clean_final.withColumn(
    # Estimate poverty rate based on GNI per capita
    "poverty_headcount_ratio",
    when(col("gni_per_capita") < 785, 70.0)  # Extreme poverty countries
    .when(col("gni_per_capita") < 1332, 50.0)  # Very poor countries
    .when(col("gni_per_capita") < 2500, 30.0)  # Poor countries
    .when(col("gni_per_capita") < 5000, 15.0)  # Lower-middle income
    .when(col("gni_per_capita") < 10000, 8.0)  # Middle income
    .when(col("gni_per_capita") < 20000, 3.0)  # Upper-middle income
    .otherwise(1.0)  # High income
).withColumn(
    # Create income category
    "income_category",
    when(col("gni_per_capita") < 1085, "Low income")
    .when(col("gni_per_capita") < 4255, "Lower-middle income")
    .when(col("gni_per_capita") < 13205, "Upper-middle income")
    .otherwise("High income")
).withColumn(
    # Estimate inequality (GINI) based on income level
    "gini_index_estimate",
    when(col("gni_per_capita") < 2000, 45.0)  # High inequality in poor countries
    .when(col("gni_per_capita") < 5000, 40.0)
    .when(col("gni_per_capita") < 10000, 37.0)
    .when(col("gni_per_capita") < 20000, 35.0)
    .otherwise(32.0)
).withColumn(
    # Create economic vulnerability score
    "economic_vulnerability",
    ((10000 - col("gni_per_capita")) / 10000 * 100)
    .cast("double")
).withColumn(
    "economic_vulnerability",
    when(col("economic_vulnerability") < 0, 0)
    .when(col("economic_vulnerability") > 100, 100)
    .otherwise(col("economic_vulnerability"))
)

# Rename to match expected downstream columns
poverty_final = economic_indicators.select(
    "country",
    "year",
    col("gni_per_capita").alias("gdp_per_capita"),  # Use GNI as GDP proxy
    "poverty_headcount_ratio",
    "gini_index_estimate",
    "income_category",
    "economic_vulnerability"
)

print(f"Economic indicators created: {poverty_final.count()} rows")

# Show distribution by income category
print("\nCountries by income category (2023):")
display(
    poverty_final
    .filter(col("year") == 2023)
    .groupBy("income_category")
    .agg(
        count("*").alias("countries"),
        round(avg("poverty_headcount_ratio"), 1).alias("avg_poverty_rate"),
        round(avg("gdp_per_capita"), 0).alias("avg_gdp_per_capita")
    )
    .orderBy("avg_gdp_per_capita")
)

# Show poorest countries
print("\nCountries with highest estimated poverty:")
display(
    poverty_final
    .filter(col("year") == 2023)
    .orderBy(desc("poverty_headcount_ratio"))
    .select("country", "gdp_per_capita", "poverty_headcount_ratio", "income_category")
    .limit(15)
)
```

```python
# Join all cleaned datasets
print("Creating unified dataset...")

from pyspark.sql.functions import *

# Standardize country names in demographics (remove regional details)
demographics_std = demographics_clean.withColumn(
    "country",
    lower(trim(regexp_replace(col("country"), ":.*", "")))
)

# Aggregate demographics by country-year (sum across all regions/subdivisions)
demographics_agg = demographics_std.groupBy("country", "year").agg(
    sum("refugee_population").alias("refugee_population")
)

print(f"Demographics aggregated: {demographics_agg.count()} country-year pairs")

# Get comprehensive list of country-year combinations
all_countries_core = (
    asylum_clean.select("country").distinct()
    .union(demographics_agg.select("country").distinct())
    .union(who_clean_final.select("country").distinct())
    .union(poverty_final.select("country").distinct())
).distinct()

years_df = spark.range(2015, 2024).select(col("id").alias("year"))

# Create base dataframe with all combinations
base_df = all_countries_core.crossJoin(years_df)
print(f"Base dataframe: {base_df.count()} country-year combinations")

# Start joining all datasets
unified_df = base_df

# Join asylum data (people fleeing from countries - crisis indicator)
print("Joining asylum data...")
unified_df = unified_df.join(
    asylum_clean,
    on=["country", "year"],
    how="left"
)

# Join demographics (refugee populations in countries)
print("Joining demographics data...")
unified_df = unified_df.join(
    demographics_agg,
    on=["country", "year"],
    how="left"
)

# Join economic/poverty data from WHO
print("Joining economic data...")
unified_df = unified_df.join(
    poverty_final,
    on=["country", "year"],
    how="left"
)

# Join health indicators from WHO (avoiding duplicate columns)
print("Joining health indicators...")
health_cols = who_clean_final.select(
    "country", "year", "life_expectancy", 
    "child_mortality_per_1000", "literacy_rate", "fertility_rate"
)
unified_df = unified_df.join(
    health_cols,
    on=["country", "year"],
    how="left"
)

# Join region mapping
print("Joining region mapping...")
unified_df = unified_df.join(
    region_mapping_complete,
    on=["country"],
    how="left"
)

# Fill missing values with intelligent defaults
print("Filling missing values...")
unified_df = unified_df.fillna({
    "refugee_applications": 0,
    "refugee_population": 0,
    "life_expectancy": 72,
    "child_mortality_per_1000": 30,
    "poverty_headcount_ratio": 15,
    "gdp_per_capita": 5000,
    "gini_index_estimate": 35,
    "economic_vulnerability": 50,
    "literacy_rate": 85,
    "fertility_rate": 2.5,
    "income_category": "Unknown",
    "region": "Unknown",
    "baseline_risk": "Unknown"
})

# Filter to keep only countries with meaningful crisis indicators
unified_df = unified_df.filter(
    (col("refugee_applications") > 0) | 
    (col("refugee_population") > 0) |
    (col("poverty_headcount_ratio") > 20) |
    (col("child_mortality_per_1000") > 40) |
    (col("gdp_per_capita") < 3000)
)

print(f"\nUnified dataset created: {unified_df.count()} rows")
print(f"Countries: {unified_df.select('country').distinct().count()}")
print(f"Years: {unified_df.select('year').distinct().count()}")

# Show data quality summary
print("\nData completeness check:")
total_rows = unified_df.count()
for col_name in ["refugee_applications", "refugee_population", "life_expectancy", 
                 "poverty_headcount_ratio", "gdp_per_capita"]:
    non_zero = unified_df.filter(col(col_name) > 0).count()
    print(f"  {col_name}: {non_zero} non-zero values ({100*non_zero/total_rows:.1f}%)")

# Display top crisis countries
print("\nTop crisis indicators (latest available data):")
display(unified_df.orderBy(desc("refugee_applications")).limit(15))
```

```python
# Calculate composite Aid Need Score
print("Calculating Aid Need Score...")

from pyspark.sql.functions import *
from pyspark.sql.window import Window

# Add normalized scores for each component
unified_scored = unified_df.withColumn(
    # Crisis displacement score (people fleeing)
    "displacement_score",
    when(col("refugee_applications") > 0, 
         log10(col("refugee_applications") + 1) * 15).otherwise(0)
).withColumn(
    # Refugee burden score (people hosted)
    "refugee_burden_score",
    when(col("refugee_population") > 0,
         log10(col("refugee_population") + 1) * 10).otherwise(0)
).withColumn(
    # Health crisis score
    "health_crisis_score",
    ((100 - col("life_expectancy")) * 0.5) + 
    (col("child_mortality_per_1000") * 0.5)
).withColumn(
    # Economic crisis score
    "economic_crisis_score",
    (col("poverty_headcount_ratio") * 1.5) + 
    (col("economic_vulnerability") * 0.5)
).withColumn(
    # Baseline risk multiplier
    "risk_multiplier",
    when(col("baseline_risk") == "Critical", 1.5)
    .when(col("baseline_risk") == "High Risk", 1.25)
    .when(col("baseline_risk") == "Medium Risk", 1.0)
    .when(col("baseline_risk") == "Low Risk", 0.75)
    .otherwise(1.0)
)

# Calculate composite Aid Need Score with weights
unified_scored = unified_scored.withColumn(
    "aid_need_score",
    (
        (col("displacement_score") * 0.30) +      # 30% - people fleeing
        (col("refugee_burden_score") * 0.20) +    # 20% - refugee hosting burden
        (col("health_crisis_score") * 0.25) +     # 25% - health indicators
        (col("economic_crisis_score") * 0.25)     # 25% - economic indicators
    ) * col("risk_multiplier")                    # Apply regional risk factor
)

# Add percentile ranking
window_percentile = Window.orderBy(desc("aid_need_score"))
unified_scored = unified_scored.withColumn(
    "aid_need_percentile",
    (percent_rank().over(window_percentile) * 100)
)

# Add crisis categories
unified_scored = unified_scored.withColumn(
    "aid_priority",
    when(col("aid_need_score") > 100, "CRITICAL")
    .when(col("aid_need_score") > 75, "URGENT")
    .when(col("aid_need_score") > 50, "HIGH")
    .when(col("aid_need_score") > 25, "MEDIUM")
    .otherwise("LOW")
)

# Add year-over-year trend analysis
window_trend = Window.partitionBy("country").orderBy("year")
unified_scored = unified_scored.withColumn(
    "score_change",
    col("aid_need_score") - lag("aid_need_score", 1).over(window_trend)
).withColumn(
    "crisis_trend",
    when(col("score_change") > 20, "Deteriorating Rapidly")
    .when(col("score_change") > 10, "Worsening")
    .when(col("score_change") > 0, "Slight Worsening")
    .when(col("score_change") < -10, "Improving")
    .when(col("score_change") < 0, "Slight Improvement")
    .otherwise("Stable")
)

# Get latest year data
latest_year = unified_scored.agg(max("year")).collect()[0][0]

print(f"Aid Need Scores calculated for {unified_scored.count()} records")
print(f"\n🚨 TOP 20 COUNTRIES NEEDING AID ({latest_year}):")

# Display top countries needing aid
top_countries = (
    unified_scored
    .filter(col("year") == latest_year)
    .select(
        "country", 
        "region",
        round("aid_need_score", 1).alias("aid_score"),
        "aid_priority",
        "crisis_trend",
        "refugee_applications",
        "refugee_population",
        round("poverty_headcount_ratio", 1).alias("poverty_rate"),
        round("life_expectancy", 1).alias("life_exp"),
        round("child_mortality_per_1000", 1).alias("child_mortality")
    )
    .orderBy(desc("aid_score"))
    .limit(20)
)

display(top_countries)

# Show distribution of aid priorities
print("\nAid Priority Distribution:")
priority_dist = (
    unified_scored
    .filter(col("year") == latest_year)
    .groupBy("aid_priority")
    .count()
    .orderBy(
        when(col("aid_priority") == "CRITICAL", 1)
        .when(col("aid_priority") == "URGENT", 2)
        .when(col("aid_priority") == "HIGH", 3)
        .when(col("aid_priority") == "MEDIUM", 4)
        .when(col("aid_priority") == "LOW", 5)
    )
)
display(priority_dist)
```
